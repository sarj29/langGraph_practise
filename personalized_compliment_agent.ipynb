```python
from typing import TypedDict, Dict
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    name : str

def compliment_me(state: AgentState) -> AgentState:
    """Simple node that adds a compliment message to the state"""
    state['name'] = "Hey, you are so cool "+ state["name"]+ "!"
    return state

graph = StateGraph(AgentState)

graph.add_node("complimentor",compliment_me)
graph.set_entry_point("complimentor")
graph.set_finish_point("complimentor")

app = graph.compile()
```

```python
import IPython
from IPython.display import Image, display
display(Image(app.get_graph().draw_mermaid_png()))
```

```python
result = app.invoke({"name":"Sreya"})
```
